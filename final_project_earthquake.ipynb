### Introduction to Global Earthquake Activity Analysis
An **earthquake** is a sudden shaking of the ground. It happens when stress that has built up inside the Earth's crust is suddenly released, usually along cracks in the rock called faults. Strong earthquakes can damage buildings, trigger tsunamis and landslides, and cause loss of life within seconds.

The purpose of this analysis is to understand where and when earthquakes occur, how strong they are (**magnitude**), and how deep they start (depth). The project uses the USGS ComCat earthquake catalog, with 8,561 worldwide earthquakes of magnitude 4.5 and above recorded in 2025. It uses Python libraries such as **Pandas**, **NumPy**, **Matplotlib**, **Seaborn** and **Plotly** to clean the data, find patterns and present the insights through clear and meaningful data visualizations.

 Data  Cleaning and Pre-processing
</h2>

### 5.1 Make column names clean and consistent
Names like `magType` and `horizontalError` mix upper and lower case. We change them to **snake_case** (`mag_type`, `horizontal_error`) so they are easy to type and read. We also rename `mag` to `magnitude`, which is clearer.

```python
# Standardizing column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(r"(?<!^)(?=[A-Z])", "_", regex=True)
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

### 5.2 Remove duplicate records
Observation: No duplicate rows or repeated earthquake event IDs were found. Therefore, no duplicate records need to be removed

### 5.3 Convert time columns to datetime
`time` and `updated` are stored as plain text. We convert them to the **datetime** type so we can find the year, month, day and hour. The values are in UTC, and we remove the time-zone label to keep things simple for charts.

```python
# Converting date-time columns
df["time"] = pd.to_datetime(df["time"], errors="coerce", utc=True).dt.tz_localize(None)
df["updated"] = pd.to_datetime(df["updated"], errors="coerce", utc=True).dt.tz_localize(None)

# Checking updated data types
df[["time", "updated"]].dtypes
```

### 5.4 Remove rows with missing core values or impossible coordinates
Five columns are the heart of every earthquake record: **time, latitude, longitude, depth and mag**. A row missing any of them is not useful, so it is removed. Rows with impossible coordinates are also removed.

```python
rows_before = len(df)

df = df.dropna(subset=["time", "latitude", "longitude", "depth", "mag"])
df = df[df["latitude"].between(-90, 90) & df["longitude"].between(-180, 180)]

print("Rows removed:", rows_before - len(df))
```

**Observation:** No missing values or invalid coordinates were found in the essential columns. Therefore, no rows were removed.

### 5.5 Handle missing values

**Text columns:** Missing values are replaced with `"Unknown"`.

**Numeric quality columns** (`nst`, `gap`, `dmin`, `rms`, error columns and `mag_nst`): Missing values are filled using the median.

The median is used because it is less affected by extreme values than the mean.

The original dataset is preserved in `df_raw` for reference.

```python
# Fill missing text values
text_cols = ["place", "mag_type", "net", "status", "location_source", "mag_source"]
df[text_cols] = df[text_cols].fillna("Unknown")

# Fill missing numeric values with median
quality_cols = ["nst", "gap", "dmin", "rms", "horizontal_error", "depth_error", "mag_error", "mag_nst"]

df[quality_cols] = df[quality_cols].fillna(df[quality_cols].median())

print("Total missing values:", df.isnull().sum().sum())
```

### 5.6 Check for outliers 
**Report observation:**

Outliers in magnitude and depth were identified using the IQR method. They were retained because extreme earthquake values are important for understanding global earthquake patterns.

```python
# Magnitude outliers
Q1 = df["mag"].quantile(0.25)
Q3 = df["mag"].quantile(0.75)
IQR = Q3 - Q1

mag_outliers = df[(df["mag"] < Q1 - 1.5 * IQR) | (df["mag"] > Q3 + 1.5 * IQR)]

print("Magnitude outliers:", len(mag_outliers))

# Depth outliers
Q1 = df["depth"].quantile(0.25)
Q3 = df["depth"].quantile(0.75)
IQR = Q3 - Q1

depth_outliers = df[(df["depth"] < Q1 - 1.5 * IQR) | (df["depth"] > Q3 + 1.5 * IQR)]

print("Depth outliers:", len(depth_outliers))
```

### 5.7 Create new (derived) columns
New columns make the analysis easier.

| New column | How it is made |
|---|---|
| `year`, `month`, `month_name`, `hour` | Taken from the `time` column |
| `depth_category` | Shallow (under 70 km), Intermediate (70 to 300 km), Deep (300 km and more) |
| `magnitude_class` | Light (4.5 to 4.9), Moderate (5.0 to 5.9), Strong (6.0 to 6.9), Major (7.0 to 7.9), Great (8.0 and above) |
| `region` | The country or area name, taken from the end of the `place` text |
| `hemisphere_ns` | Northern or Southern from latitude and longitude |

```python
# Date and time features
df["year"] = df["time"].dt.year
df["month"] = df["time"].dt.month
df["month_name"] = df["time"].dt.month_name()
df["hour"] = df["time"].dt.hour
```

```python
# Depth category
df["depth_category"] = pd.cut(
    df["depth"],
    bins=[-np.inf, 70, 300, np.inf],
    right=False,
    labels=["Shallow", "Intermediate", "Deep"]
)
```

```python
# Magnitude class
df["magnitude_class"] = pd.cut(
    df["mag"],
    bins=[-np.inf, 5, 6, 7, 8, np.inf],
    right=False,
    labels=["Light", "Moderate", "Strong", "Major", "Great"]
)
```

```python
#region
df["region"] = df["place"].str.split(",").str[-1].str.strip()
```

```python
# Checking the most common regions
df["region"].value_counts().head(15)
```

```python
# Northern or Southern hemisphere
df["hemisphere_ns"] = np.where(
    df["latitude"] >= 0, "Northern", "Southern"
)
```
